```python
import faiss
import numpy as np
from sentence_transformers import SentenceTransformer # for Embedding
```

#### load sentences into embedding model


```python
model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
# Sample documents
docs = [
    "Artificial Intelligence is changing the world",
    "Machine learning improves software",
    "Cats are lovely animals",
    "AI will transform industries"
]
```

### create sentence embedding

```python
doc_embedding=model.encode(docs)
```

##### This is part of using FAISS (Facebook AI Similarity Search) to create a vector index and add document embeddings to it so you can search for similar vectors later.



### dimension = doc_embeddings.shape[1]
Definition:
This line gets the number of dimensions of each embedding vector.

doc_embeddings is assumed to be a 2D NumPy array (e.g., shape [num_docs, embedding_dim]).

shape[1] gives the number of columns = embedding dimension.

Example:
If doc_embeddings.shape is (100, 768), then dimension = 768.

### index = faiss.IndexFlatL2(dimension)


Definition:
This creates a FAISS index using the L2 (Euclidean) distance for similarity search.

IndexFlatL2: A simple, flat index that compares all vectors using L2 distance (squared Euclidean).

dimension: Must match the dimension of the vectors you’ll add.

🧠 Note: IndexFlatL2 is fast for small datasets but not optimized for large-scale search (millions of vectors). For large data, use IVFFlat or HNSW.

### index.add(np.array(doc_embeddings))


Definition:
This adds the embedding vectors to the FAISS index.



```python
# Step 2: Create FAISS index and add vectors

dimensions=doc_embedding.shape[1]
index=faiss.IndexFlatL2(dimensions)
index.add(np.array(doc_embedding))
```

 ### query_embedding = model.encode([query])


 You're converting the query text into a vector (embedding) using a pre-trained sentence embedding model


 ### D, I = index.search(np.array(query_embedding), k=2)


This line performs the similarity search on the FAISS index.

query_embedding: Query vector.

k=2: You want the top 2 most similar document vectors.

index.search(...) returns:

```python
 # Step 3: Search for similar sentences to a query

query="how AI transforming society ?"
query_embedding=model.encode([query]) # text to embedding
D, I = index.search(np.array(query_embedding), k=2)
# search in the document to get accurate result of the query and give top 2 of it

```

show most similar sentence

```python
print(" Vector DB - Top matching Sentences.....")
for i in I[0]:
  print(f" -{docs[i]}")
```
