```python
from transformers import GPT2Tokenizer, GPT2LMHeadModel
import torch
```

#### This code loads a pretrained GPT-2 language model and its tokenizer from Hugging Face’s transformers library.



```python
tokenizer=GPT2Tokenizer.from_pretrained("gpt2")
model=GPT2LMHeadModel.from_pretrained("gpt2")
```

**encode()** turns the input text into a list of token IDs (integers that represent words or subwords).

These IDs are the actual input that GPT-2 understands.

```python
input_text = "Cristiano Ronaldo is the best football"

input_ids=tokenizer.encode(input_text,return_tensors="pt")
```

| Parameter                | Meaning                                                                                                              |
| ------------------------ | -------------------------------------------------------------------------------------------------------------------- |
| `input_ids=input_ids`    | This is the **tokenized input** (e.g., `"Once upon a time"`), passed as model input.                                 |
| `max_length=50`          | Maximum total **length of the output** (including the input prompt).                                                 |
| `num_return_sequences=1` | Number of different **generated variations** you want.                                                               |
| `do_sample=True`         | Enables **sampling** instead of greedy decoding, which makes output **more creative/random**.                        |
| `temperature=0.8`        | Controls **randomness** in sampling: <br>🔹 Lower = safer/more predictable <br>🔹 Higher = more random and creative. |


```python
# Generate text continuation
output = model.generate(
    input_ids=input_ids,
    max_length=200,           # total length including input
    num_return_sequences=1,  # how many outputs to return
    do_sample=True,          # enable sampling for creative results
    temperature=0.8          # controls randomness (lower = more focused)
)
```

# Decode output tokens to text

✅ The final human-readable string decoded from the model's generated output tokens (after removing special tokens like <|endoftext|>).

```python
generated_text=tokenizer.decode(output[0],skip_special_tokens=True)
```
